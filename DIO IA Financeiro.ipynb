```python
historico = []

print("=== Guia Financeiro IA ===")

while True:
    pergunta = input("\nDigite sua pergunta (ou 'sair'): ")

    if pergunta.lower() == "sair":
        break

    historico.append(pergunta)

    if "cdi" in pergunta.lower():
        resposta = "CDI é uma taxa usada como referência para diversos investimentos."

    elif "tesouro" in pergunta.lower():
        resposta = "Tesouro Direto é um programa para investir em títulos públicos."

    elif "cartão" in pergunta.lower():
        resposta = "Cartão de crédito permite compras com pagamento posterior."

    elif "empréstimo" in pergunta.lower():
        resposta = "Empréstimo é um valor recebido que deverá ser devolvido com juros."

    elif "juros compostos" in pergunta.lower():
        resposta = "Juros compostos são calculados sobre o valor inicial mais os juros acumulados."

    else:
        resposta = "Desculpe, ainda não tenho informação sobre esse assunto."

    print("\nIA:", resposta)

print("\nHistórico da conversa:")
for item in historico:
    print("-", item)
```

```python
print("=== Simulador Financeiro ===")

valor = float(input("Valor investido: R$ "))
taxa = float(input("Taxa de juros (% ao mês): "))
meses = int(input("Quantidade de meses: "))

montante = valor * ((1 + taxa/100) ** meses)

print(f"\nValor inicial: R$ {valor:.2f}")
print(f"Valor final: R$ {montante:.2f}")
print(f"Lucro: R$ {montante - valor:.2f}")
```
